# Project 27 — Gradient Boosting

## Step1 — Load the Data

```python
import pandas as pd

data = {
    "Age": [22, 25, 47, 52, 46, 56, 27, 30, 40, 35],
    "Salary": [15000, 29000, 48000, 60000, 52000, 75000, 32000, 58000, 62000, 45000],
    "Purchased": [0, 0, 1, 1, 1, 1, 0, 1, 1, 0]
}

df_tree = pd.DataFrame(data)
```

## Step 1 — Train Gradient Boosting Model

```python
from sklearn.ensemble import GradientBoostingClassifier
import pandas as pd

# Features & Target
X = df_tree[["Age", "Salary"]]
y = df_tree["Purchased"]

# Train model
gb = GradientBoostingClassifier(random_state=42)

gb.fit(X, y)

# Predict
pred = gb.predict(
    pd.DataFrame([[30, 58000]], columns=["Age", "Salary"])
)

print("Gradient Boosting Prediction:", pred[0])
```

## Step 2 — Feature Importance in Gradient Boosting

```python
print("Age importance:", gb.feature_importances_[0])
print("Salary importance:", gb.feature_importances_[1])
```
