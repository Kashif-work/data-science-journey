# Project 12_Feature Scaling

```python
# import library
import pandas as pd

# Load the data
df = pd.read_csv("final_data.csv")
print(df)
```

```python
from sklearn.preprocessing import StandardScaler

# Features
X = df[["Age", "Salary"]]

# Scaler
scaler = StandardScaler()

# Fit and transform
X_scaled = scaler.fit_transform(X)

print("Scaled Data:")
print(X_scaled)
```

```python
from sklearn.linear_model import LogisticRegression

# Without scaling
clf1 = LogisticRegression(max_iter=1000)
clf1.fit(df[["Age", "Salary"]], df["Salary_Category"])

pred1 = clf1.predict(
    pd.DataFrame([[30, 60000]], columns=["Age", "Salary"])
)


# With scaling
from sklearn.preprocessing import StandardScaler

X = df[["Age", "Salary"]]
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

clf2 = LogisticRegression(max_iter=1000)
clf2.fit(X_scaled, df["Salary_Category"])

scaled_input = scaler.transform(
    pd.DataFrame([[30, 60000]], columns=["Age", "Salary"])
)

pred2 = clf2.predict(scaled_input)

print("Without Scaling:", pred1[0])
print("With Scaling:", pred2[0])
```

```python
# save data
df.to_csv("final_data.csv", index=False)
```
