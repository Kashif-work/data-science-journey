# Project 16_Gradient Boosting

```python
# import library
import pandas as pd

# Load the data
df = pd.read_csv("final_data.csv")
print(df)
```

```python
from sklearn.ensemble import GradientBoostingClassifier
import pandas as pd

# Features & target
X = df[["Age", "Salary"]]
y = df["Salary_Category"]

# Model
gb = GradientBoostingClassifier(random_state=42)

# Train
gb.fit(X, y)

# Predict
prediction = gb.predict(
    pd.DataFrame([[30, 60000]], columns=["Age", "Salary"])
)

print("Gradient Boosting Prediction:", prediction[0])
```

```python
# save data
df.to_csv("final_data.csv", index=False)
```
