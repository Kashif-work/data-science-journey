# Project 24 — Real ML Workflow (missing values + encoding + model)

## Step 1 — Create Realistic Dataset

```python
import pandas as pd
import numpy as np

data = {
    "Age": [22, 25, np.nan, 35, 40, 28, 50, np.nan],
    "Salary": [25000, 32000, 45000, 58000, 62000, np.nan, 80000, 52000],
    "City": ["Lahore", "Karachi", "Lahore", "Islamabad", "Karachi", "Lahore", "Islamabad", "Karachi"],
    "Purchased": [0, 0, 1, 1, 1, 0, 1, 1]
}

df_real_ml = pd.DataFrame(data)

print(df_real_ml)
```

## Step 2 — Handle Missing Values

```python
# Fill missing values with mean
df_real_ml["Age"] = df_real_ml["Age"].fillna(df_real_ml["Age"].mean())

df_real_ml["Salary"] = df_real_ml["Salary"].fillna(df_real_ml["Salary"].mean())

print(df_real_ml)
```

## Step 3 — Encode Categorical Data

```python
df_encoded = pd.get_dummies(df_real_ml, columns=["City"])

print(df_encoded)
```

## Step 4 — Train Real ML Model

```python
from sklearn.linear_model import LogisticRegression

# Features & Target
X = df_encoded.drop("Purchased", axis=1)
y = df_encoded["Purchased"]

# Train model
model = LogisticRegression(max_iter=1000)
model.fit(X, y)

# Predict new customer
new_customer = pd.DataFrame(
    [[30, 55000, 0, 1, 0]],
    columns=X.columns
)

pred = model.predict(new_customer)

print("Prediction:", pred[0])
```
