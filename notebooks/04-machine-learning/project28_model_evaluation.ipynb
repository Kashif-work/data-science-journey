# Project 28 — Model Evaluation

## First create Dataset

```python
import pandas as pd

data = {
    "Age": [22, 25, 47, 52, 46, 56, 27, 30, 40, 35],
    "Salary": [15000, 29000, 48000, 60000, 52000, 75000, 32000, 58000, 62000, 45000],
    "Purchased": [0, 0, 1, 1, 1, 1, 0, 1, 1, 0]
}

df_tree = pd.DataFrame(data)

print(df_tree)
```

## Step 2 — Train/Test Split

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier

# Features & Target
X = df_tree[["Age", "Salary"]]
y = df_tree["Purchased"]

# Split data
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.3,
    random_state=42
)

# Train model
rf = RandomForestClassifier(random_state=42)

rf.fit(X_train, y_train)

# Predict on TEST data
y_pred = rf.predict(X_test)

print("Predictions:", y_pred)
print("Actual:", y_test.values)
```

## Step 2 — Calculate Accuracy

```python
from sklearn.metrics import accuracy_score

acc = accuracy_score(y_test, y_pred)

print("Accuracy:", acc)
```

## Step 3 — Confusion Matrix

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, y_pred)

print(cm)
```

## Step 4 — Precision, Recall, F1-Score 

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred))
```
