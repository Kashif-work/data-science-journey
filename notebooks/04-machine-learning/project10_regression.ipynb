# Project 10_Regression

```python
# import library
import pandas as pd

# Load the data
df = pd.read_csv("final_data.csv")
print(df)
```

```python
from sklearn.linear_model import LinearRegression
import numpy as np

# Prepare data
X = df[["Age"]]   # feature
y = df["Salary"]  # target

# Create model
model = LinearRegression()

# Train model
model.fit(X, y)

# Predict salary for age 30
predicted_salary = model.predict(pd.DataFrame([[30]], columns=["Age"]))

print("Predicted Salary for Age 30:", predicted_salary[0])
```

```python
# Model parameters

print("Slope (Coefficient):", model.coef_[0])
print("Intercept:", model.intercept_)
```

```python
# Model evaluation (R² score)

r2 = model.score(X, y)

print("R² Score:", r2)
```

```python
from sklearn.linear_model import LogisticRegression

# Features & target
X = df[["Age"]]
y = df["Salary_Category"]

# Model
clf = LogisticRegression()

# Train
clf.fit(X, y)

# Predict for age 30
prediction = clf.predict(pd.DataFrame([[30]], columns=["Age"]))

print("Predicted Salary Category for Age 30:", prediction[0])
```

```python
# Step 2: Prediction Probabilities

probabilities = clf.predict_proba(pd.DataFrame([[30]], columns=["Age"]))

print("Probabilities:", probabilities)
print("Classes:", clf.classes_)
```

```python
# Step 3 — Model Accuracy
from sklearn.metrics import accuracy_score

# Predictions on training data
y_pred = clf.predict(X)

# Accuracy
accuracy = accuracy_score(y, y_pred)

print("Accuracy:", accuracy)
```
