```python
# Project 3 - Pandas Introduction

import pandas as pd

# Create dataset manually
data = {
    "Name": ["Ali", "Sara", "John", "Ayesha", "David"],
    "Age": [25, 30, 22, 28, 35],
    "Salary": [50000, 60000, 45000, 52000, 80000]
}

df = pd.DataFrame(data)

print(df)
```

```python
# Step 2: Explore the DataFrame

print("First 3 rows:")
print(df.head(3))
```

```python
print("\nData Info:")
print(df.info())
```

```python
print("\nSummary Statistics:")
print(df.describe())
```

```python
# Step 3: Selecting Data

# Select one column
print("Names:")
print(df["Name"])

# Select multiple columns
print("\nName and Salary:")
print(df[["Name", "Salary"]])

# Select rows using index
print("\nFirst 2 rows:")
print(df.iloc[0:2])
```

```python
# Step 4: Filtering Data

# Filter people with Age > 25
filtered_df = df[df["Age"] > 25]

print("People with Age > 25:")
print(filtered_df)
```

```python
# Step 5: Create New Column

# Add bonus = 10% of salary
df["Bonus"] = df["Salary"] * 0.10

print("Data with Bonus:")
print(df)
```

```python
# Step 6: Sort Data

# Sort by Salary (highest first)
sorted_df = df.sort_values(by="Salary", ascending=False)

print("Sorted by Salary (High to Low):")
print(sorted_df)
```

```python
# save the data
import pandas as pd

df.to_csv("employee_data.csv", index=False)
```
