```python
# import library
import pandas as pd

# Load the data
df = pd.read_csv("employee_salary_dataset.csv")
print(df)
```

```python
import matplotlib.pyplot as plt

# Bar chart of Salary by Name
plt.bar(df["Name"], df["Salary"])

plt.title("Salary by Employee")
plt.xlabel("Name")
plt.ylabel("Salary")

plt.show()
```

```python
# Step 2: Salary Distribution

plt.hist(df["Salary"])

plt.title("Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("Frequency")

plt.show()
```

```python
# Step 3: Box Plot

plt.boxplot(df["Salary"])

plt.title("Salary Box Plot")
plt.ylabel("Salary")

plt.show()
```

```python
# save data
df.to_csv("employee_salary_dataset", index=False)
```
