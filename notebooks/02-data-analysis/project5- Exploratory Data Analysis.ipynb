# Project 5 - Exploratory Data Analysis

```python
# Load the data
import pandas as pd

df = pd.read_csv("employee_salary_dataset.csv")
print(df)
```

```python
# Average salary
avg_salary = df["Salary"].mean()

# Maximum salary
max_salary = df["Salary"].max()

# Minimum salary
min_salary = df["Salary"].min()

print("Average Salary:", avg_salary)
print("Max Salary:", max_salary)
print("Min Salary:", min_salary)
```

```python
# Step 2: Group Analysis

# Average salary by Age
grouped = df.groupby("Age")["Salary"].mean()

print("Average Salary by Age:")
print(grouped)
```

```python
# Step 3: Top Earners

top_earners = df.sort_values(by="Salary", ascending=False).head(3)

print("Top 3 Earners:")
print(top_earners)
```

```python
df[df["Salary"] > df["Salary"].mean()][["Name", "Salary"]]
```
