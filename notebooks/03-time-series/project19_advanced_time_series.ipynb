# Project 19: Advanced Time Series

```python
# import library
import pandas as pd

# Load the data
df_ts = pd.read_csv("customer_sales_dataset.csv")
print(df_ts)
```

```python
# Moving Average (window = 2)
df_ts["MA_2"] = df_ts["Sales"].rolling(window=2).mean()

print(df_ts)
```

```python
df_ts[["Sales", "MA_2"]].plot()

import matplotlib.pyplot as plt
plt.title("Original vs Moving Average")
plt.xlabel("Time")
plt.ylabel("Sales")
plt.show()
```

```python
# Moving Average with window = 3
df_ts["MA_3"] = df_ts["Sales"].rolling(window=3).mean()

# Plot all
df_ts[["Sales", "MA_2", "MA_3"]].plot()

import matplotlib.pyplot as plt
plt.title("Sales vs MA(2) vs MA(3)")
plt.xlabel("Time")
plt.ylabel("Sales")
plt.show()
```

```python
# Last MA_3 value
last_ma3 = df_ts["MA_3"].iloc[-1]

print("Predicted next value (MA_3):", last_ma3)
```

```python
# save the data
df_ts.to_csv("customer_sales_dataset.csv", index=False)
```
