# Project 18 __ Time Series Forecasting

```python
# Create Time-Based Data
import pandas as pd

data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May"],
    "Sales": [200, 220, 250, 270, 300]
}

df_ts = pd.DataFrame(data)

print(df_ts)
```

```python
# Convert to Real Time Format
df_ts["Month"] = pd.to_datetime(df_ts["Month"], format="%b")

df_ts.set_index("Month", inplace=True)

print(df_ts)
```

```python
# Visualize Time Trend
import matplotlib.pyplot as plt

df_ts["Sales"].plot()

plt.title("Sales Over Time")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

```python
# Simple Forecast

import numpy as np

# Convert index to numbers
df_ts["Time"] = np.arange(len(df_ts))

# Train simple model
from sklearn.linear_model import LinearRegression

X = df_ts[["Time"]]
y = df_ts["Sales"]

model = LinearRegression()
model.fit(X, y)

# Predict next month (June → index = 5)
pred = model.predict(pd.DataFrame([[5]], columns=["Time"]))

print("Predicted Sales for next month:", pred[0])
```
