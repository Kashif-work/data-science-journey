# Project 30 — Real Dataset ML Workflow

## Step 1 — Create Realistic Dataset

```python
import pandas as pd

data = {
    "Age": [22, 25, 47, 52, 46, 56, 27, 30, 40, None, 35, 48],
    "Salary": [25000, 32000, 48000, 60000, 52000, 75000, None, 58000, 62000, 54000, 45000, 80000],
    "Experience": [1, 2, 10, 12, 9, 15, 3, 4, 8, 7, 5, 14],
    "City": ["Lahore", "Karachi", "Lahore", "Islamabad", "Karachi",
             "Islamabad", "Lahore", "Karachi", "Islamabad",
             "Karachi", "Lahore", "Islamabad"],
    "Purchased": [0, 0, 1, 1, 1, 1, 0, 1, 1, 1, 0, 1]
}

df_real = pd.DataFrame(data)

print(df_real)
```

## Step 2 — Handle Missing Values Properly

```python
# Fill missing Age with mean
df_real["Age"] = df_real["Age"].fillna(
    df_real["Age"].mean()
)

# Fill missing Salary with mean
df_real["Salary"] = df_real["Salary"].fillna(
    df_real["Salary"].mean()
)

print(df_real)
```

## Step 3 — Encode Categorical Data

```python
df_encoded = pd.get_dummies(
    df_real,
    columns=["City"]
)

print(df_encoded)
```

## Step 4 — Train/Test Split + Model Training

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier

# Features & Target
X = df_encoded.drop("Purchased", axis=1)
y = df_encoded["Purchased"]

# Split
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.3,
    random_state=42
)

# Train model
model = RandomForestClassifier(random_state=42)

model.fit(X_train, y_train)

# Predictions
y_pred = model.predict(X_test)

print("Predictions:", y_pred)
print("Actual:", y_test.values)
```

## Step 5 — Evaluate the Real Model

```python
from sklearn.metrics import accuracy_score
from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report

# Accuracy
acc = accuracy_score(y_test, y_pred)

# Confusion Matrix
cm = confusion_matrix(y_test, y_pred)

# Report
report = classification_report(y_test, y_pred)

print("Accuracy:", acc)
print("\nConfusion Matrix:\n", cm)
print("\nClassification Report:\n", report)
```
