# Project 25 — Decision Trees

## Step 1 — Create Dataset

```python
import pandas as pd

data = {
    "Age": [22, 25, 47, 52, 46, 56, 27, 30, 40, 35],
    "Salary": [15000, 29000, 48000, 60000, 52000, 75000, 32000, 58000, 62000, 45000],
    "Purchased": [0, 0, 1, 1, 1, 1, 0, 1, 1, 0]
}

df_tree = pd.DataFrame(data)

print(df_tree)
```

## Step 2 — Train Decision Tree

```python
from sklearn.tree import DecisionTreeClassifier

# Features & Target
X = df_tree[["Age", "Salary"]]
y = df_tree["Purchased"]

# Train model
tree_model = DecisionTreeClassifier(random_state=42)

tree_model.fit(X, y)

# Predict
import pandas as pd

pred = tree_model.predict(
    pd.DataFrame([[30, 58000]], columns=["Age", "Salary"])
)

print("Prediction:", pred[0])
```

## Step 3 — Visualize Tree Logic

```python
from sklearn import tree
import matplotlib.pyplot as plt

plt.figure(figsize=(10,6))

tree.plot_tree(
    tree_model,
    feature_names=["Age", "Salary"],
    class_names=["No", "Yes"],
    filled=True
)

plt.show()
```

## Step 4 — Limit Tree Depth

```python
# Smaller/simple tree
small_tree = DecisionTreeClassifier(
    max_depth=1,
    random_state=42
)

small_tree.fit(X, y)

pred_small = small_tree.predict(
    pd.DataFrame([[30, 58000]], columns=["Age", "Salary"])
)

print("Prediction with limited depth:", pred_small[0])
```

## Step 5 — Feature Importance

```python
print("Age importance:", tree_model.feature_importances_[0])
print("Salary importance:", tree_model.feature_importances_[1])
```
