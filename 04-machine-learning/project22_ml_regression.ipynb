## Project 22_ML Regression

## Step 1 — Create House Price Datase

```python
import pandas as pd

data = {
    "Size_sqft": [500, 800, 1200, 1500, 1800, 2000, 2200, 2500],
    "Price": [100000, 150000, 220000, 260000, 300000, 340000, 360000, 400000]
}

df_reg = pd.DataFrame(data)

print(df_reg)
```

## Step 2 — Train Regression Model

```python
from sklearn.linear_model import LinearRegression

# Features & Target
X = df_reg[["Size_sqft"]]
y = df_reg["Price"]

# Train model
model = LinearRegression()
model.fit(X, y)

# Predict price for 1700 sqft house
import pandas as pd

pred = model.predict(pd.DataFrame([[1700]], columns=["Size_sqft"]))

print("Predicted Price:", pred[0])
```

## Step 3 — Understand Model Equation

```python
print("Slope:", model.coef_[0])
print("Intercept:", model.intercept_)
```

## Step 4 — Evaluate Regression Model

```python
from sklearn.metrics import r2_score

y_pred = model.predict(X)

r2 = r2_score(y, y_pred)

print("R² Score:", r2)
```
