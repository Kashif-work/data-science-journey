# Project 11_Classification

```python
# import library
import pandas as pd

# Load the data
df = pd.read_csv("final_data.csv")
print(df)
```

```python
from sklearn.linear_model import LogisticRegression

# Features & target
X = df[["Age"]]
y = df["Salary_Category"]

# Model
clf = LogisticRegression()

# Train
clf.fit(X, y)

# Predict for age 30
prediction = clf.predict(pd.DataFrame([[30]], columns=["Age"]))

print("Predicted Salary Category for Age 30:", prediction[0])
```

```python
# Step 2: Prediction Probabilities

probabilities = clf.predict_proba(pd.DataFrame([[30]], columns=["Age"]))

print("Probabilities:", probabilities)
print("Classes:", clf.classes_)
```

```python
# Step 3 — Model Accuracy
from sklearn.metrics import accuracy_score

# Predictions on training data
y_pred = clf.predict(X)

# Accuracy
accuracy = accuracy_score(y, y_pred)

print("Accuracy:", accuracy)
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y, y_pred)

print("Confusion Matrix:")
print(cm)
```

```python
from sklearn.metrics import classification_report

report = classification_report(y, y_pred)

print("Classification Report:")
print(report)
```

```python
# save data
df.to_csv("final_data.csv", index=False)
```
