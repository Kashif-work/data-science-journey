# Project 26 — Random Forest

## Step-1 Load the data

```python
import pandas as pd

data = {
    "Age": [22, 25, 47, 52, 46, 56, 27, 30, 40, 35],
    "Salary": [15000, 29000, 48000, 60000, 52000, 75000, 32000, 58000, 62000, 45000],
    "Purchased": [0, 0, 1, 1, 1, 1, 0, 1, 1, 0]
}

df_tree = pd.DataFrame(data)
```

## Step 2 — Train Random Forest

```python
from sklearn.ensemble import RandomForestClassifier
import pandas as pd

# Features & Target
X = df_tree[["Age", "Salary"]]
y = df_tree["Purchased"]

# Train Random Forest
rf = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

rf.fit(X, y)

# Predict
pred = rf.predict(
    pd.DataFrame([[30, 58000]], columns=["Age", "Salary"])
)

print("Random Forest Prediction:", pred[0])
```

## Step 3 — Feature Importance in Random Forest

```python
print("Age :", rf.feature_importances_[0])
print("Salary :", rf.feature_importances_[1])
```
