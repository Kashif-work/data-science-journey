# Project 23: Multi-Feature Regression (Realistic ML)

## Step 1 — Create Better Dataset

```python
import pandas as pd

data = {
    "Size_sqft": [500, 800, 1200, 1500, 1800, 2000, 2200, 2500],
    "Bedrooms": [1, 2, 3, 3, 4, 4, 5, 5],
    "Price": [100000, 150000, 220000, 260000, 300000, 340000, 360000, 400000]
}

df_multi = pd.DataFrame(data)

print(df_multi)
```

## Step 2 — Train Multi-Feature Model

```python
from sklearn.linear_model import LinearRegression

# Features & Target
X = df_multi[["Size_sqft", "Bedrooms"]]
y = df_multi["Price"]

# Train model
model = LinearRegression()
model.fit(X, y)

# Predict price
import pandas as pd
pred = model.predict(pd.DataFrame([[1700, 3]], columns=["Size_sqft", "Bedrooms"]))

print("Predicted Price:", pred[0])
```

## Step 3 — Feature Importance (CRITICAL)

```python
print("Size coefficient:", model.coef_[0])
print("Bedrooms coefficient:", model.coef_[1])
```

## Step 4 — Evaluate Multi-Feature Model

```python
from sklearn.metrics import r2_score

y_pred = model.predict(X)

r2 = r2_score(y, y_pred)

print("R² Score:", r2)
```
