# Project 13_Feature Selection

```python
# import library
import pandas as pd

# Load the data
df = pd.read_csv("final_data.csv")
print(df)
```

```python
# Correlation with Salary
corr = df.corr(numeric_only=True)

print("Correlation Matrix:")
print(corr)
```

```python
from sklearn.linear_model import LinearRegression

# Features
X = df[["Age"]]
y = df["Salary"]

model = LinearRegression()
model.fit(X, y)

print("Feature Importance (Coefficient):", model.coef_[0])
```

```python
# save data
df.to_csv("final_data.csv", index=False)
```
