# Project 14_ Decision Trees

###  Step: 1 — Load the data

```python
# import library
import pandas as pd

# Load the data
df = pd.read_csv("final_data.csv")
print(df)
```

###  Step: 2 — Build Decision Tree

```python
from sklearn.tree import DecisionTreeClassifier

# Features & target
X = df[["Age", "Salary"]]
y = df["Salary_Category"]

# Model
tree = DecisionTreeClassifier(random_state=42)

# Train
tree.fit(X, y)

# Predict
import pandas as pd

prediction = tree.predict(
    pd.DataFrame([[30, 60000]], columns=["Age", "Salary"])
)

print("Prediction:", prediction[0])
```

## Step 3 — Understand Tree Logic

```python
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt

plt.figure(figsize=(8,6))
plot_tree(tree, feature_names=["Age", "Salary"], class_names=tree.classes_, filled=True)
plt.show()
```

### Step 3 — Overfitting

```python
tree = DecisionTreeClassifier(random_state=42, max_depth=1)
tree.fit(X, y)

prediction = tree.predict(
    pd.DataFrame([[30, 60000]], columns=["Age", "Salary"])
)

print("Prediction with limited depth:", prediction[0])
```
