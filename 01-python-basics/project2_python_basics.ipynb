#  Project 2 - Python Basics

```python
# Step 1: Variables
name = "Kashif"
age = 22

print("Name:", name)
print("Age:", age)

# Step 2: List
numbers = [10, 20, 30, 40, 50]
print("Numbers:", numbers)

# Step 3: Loop
for num in numbers:
    print("Number:", num)

# Step 4: Function
def square(x):
    return x * x

print("Square of 5:", square(5))
```

```python
# Step 5: Apply function to list

squared_numbers = []

for num in numbers:
    squared_numbers.append(square(num))

print("Squared Numbers:", squared_numbers)
```

```python
# Step 6: Filter numbers greater than 20

filtered_numbers = []

for num in numbers:
    if num > 20:
        filtered_numbers.append(num)

print("Filtered Numbers:", filtered_numbers)
```

```python
# Step 7: Filter and transform together

result = []

for num in numbers:
    if num > 20:
        result.append(square(num))

print("Filtered + Squared:", result)
```

```python
# Step 8: Using list comprehension

result = [square(num) for num in numbers if num > 20]

print("Optimized Result:", result)
```

```python
result = [num * 2 for num in numbers if num < 40]
print("Result:", result)
```
