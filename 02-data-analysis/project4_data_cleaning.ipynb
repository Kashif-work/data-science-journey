# Project 4 - Data Cleaning

```python
import pandas as pd

# Create messy dataset
data = {
    "Name": ["Ali", "Sara", "John", "Ayesha", "David", "Ali"],
    "Age": [25, None, 22, 28, 35, 25],
    "Salary": [50000, 60000, None, 52000, 80000, 50000]
}

df = pd.DataFrame(data)

print("Original Data:")
print(df)
```

```python
# Step 2: Handle Missing Values

# Fill missing Age with average age
df["Age"] = df["Age"].fillna(df["Age"].mean())

# Fill missing Salary with average salary
df["Salary"] = df["Salary"].fillna(df["Salary"].mean())

print("After Handling Missing Values:")
print(df)
```

```python
# Step 3: Remove Duplicates

df = df.drop_duplicates()

print("After Removing Duplicates:")
print(df)
```

```python
# Step 4: Fix Data Types

df["Age"] = df["Age"].astype(int)
df["Salary"] = df["Salary"].astype(int)

print("After Fixing Data Types:")
print(df)

print("\nData Types:")
print(df.dtypes)
```

```python
# save data
df.to_csv("employee_salary_dataset.csv", index=False)
```
