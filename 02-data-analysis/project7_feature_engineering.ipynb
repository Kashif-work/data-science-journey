# Project 7_Feature Engineering

```python
# import library
import pandas as pd

# Load the data
df = pd.read_csv("employee_salary_dataset.csv")
print(df)
```

```python
# Create Salary Category
def categorize_salary(salary):
    if salary < 55000:
        return "Low"
    elif salary < 70000:
        return "Medium"
    else:
        return "High"

df["Salary_Category"] = df["Salary"].apply(categorize_salary)

print("Data with Salary Category:")
print(df)
```

```python
# Step 2: Age Group Feature

def age_group(age):
    if age < 25:
        return "Young"
    elif age < 30:
        return "Mid"
    else:
        return "Senior"

df["Age_Group"] = df["Age"].apply(age_group)

print("Data with Age Group:")
print(df)
```

```python
# Step 3: Combine Features

df["Profile"] = df["Age_Group"] + "_" + df["Salary_Category"]

print("Data with Profile Feature:")
print(df)
```

```python
# save data
df.to_csv("employee_salary_dataset.csv", index=False)
```
